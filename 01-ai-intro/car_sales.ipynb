```python
import pandas as pd
df = pd.read_csv('../data/car_sales_cleaned1.csv')
df
```

```python
df = df.drop_duplicates() # removing duplicates
df.shape
```

```python
df = df.dropna() # removing rows with missing values
df.shape
```

```python
valuesColModel = df['Model'].unique()
valuesColModel
```

```python
dfProjection = df[['Annual Income', 'Price ($)']]
dfProjection
```

```python
from sklearn.tree import DecisionTreeRegressor
model = DecisionTreeRegressor()
x = dfProjection[['Annual Income']]
y = dfProjection[['Price ($)']]
model.fit(x.values, y)
prediction = model.predict([[30000]])
prediction
```
