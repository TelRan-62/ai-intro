```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
import converter

df = pd.read_csv('../data/car_sales_cleaned1.csv')

df = df.drop_duplicates()  #removing duplicates
df = df.dropna()  # removing rows with missing data
df
```

```python
model = DecisionTreeClassifier()
y = df[['Model']]
mapper = converter.columns_mapper(['Gender', 'Company', 'Model', 'Transmission', 'Color'], df)
dfConverted = converter.convert_x(df, mapper)
X = dfConverted[['Gender', 'Color']]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
model.fit(X_train, y_train)
predictions = model.predict(X_test)
score = accuracy_score(y_test, predictions)
score
```

```python
import joblib

model.fit(X.values, y)
predictions = model.predict([[mapper['Gender']['Female'], mapper['Color']['Pale White']]])
predictions
```

```python
joblib.dump(model, '../models/car_gender_model_hw.joblib')
```
